# Stellohedron family — full exploration, n = 1, 2, 3, 4

For each n (the stellohedron of the star graph on n+1 nodes — 1 center, n leaves — has dimension n, matching `stellohedron.sage`), this notebook:

1. lists all vertices (Devadoss's graph-associahedron coordinates, reduced to the full-dimensional chart — see `reduce_codim1`),
2. computes the canonical form via the general nbc method (Brown–Dupont Prop. 6.7), checking the defining pole-structure property,
3. computes the projective (polar) dual,
4. checks the **volume conjecture** at the centroid (see `simplex_explorer.ipynb`'s n=1 section for why it has to be the centroid),
5. enumerates all triangulations and identifies which are regular (n = 1, 2 only — see below),
6. computes the secondary polytope and its vertex embedding (n = 1, 2 only).

**Triangulation enumeration stops at n=2.** n=3 (16 points) was measured to exceed 90 seconds while building this notebook, vs. well under a second for n=1, 2 — the same pattern already seen on the associahedron and cyclohedron. Canonical form computation reaches a bit further (n=4 took under 5s; n=5 exceeded 60s), so n=3 and n=4 still get the canonical form, dual, and volume conjecture, just not the triangulation/secondary-polytope section. The stellohedron's vertex count (OEIS [A000522](https://oeis.org/A000522): 2, 5, 16, 65, 326, ...) grows a little faster than the cyclohedron's for the same n, which is why this notebook's limits sit one step lower throughout.

The vertex generator here (`stellohedron_vertices`, in `common.sage`) is a specific case of the same general **graph-associahedron** construction (Devadoss, [arXiv:math/0612530](https://arxiv.org/abs/math/0612530)) used for the cyclohedron — see `graph-associahedra.md` on the site and `cyclohedron_explorer.ipynb` in this folder.

**Requires the `sagemath` Jupyter kernel** and must be opened from the same synced folder as the `.sage` files — see `README.md`.

```sage
load("general_canonical_forms.sage")
```

That `load` pulls in `common.sage` (vertex generators, `reduce_codim1`, `polar_dual`, `secondary_polytope_data`, `canonical_form_by_vertex`) and `vertex_sum_canonical_forms.sage` too, and runs `general_canonical_forms.sage`'s own test suite as a side effect (scroll up for that PASS/FAIL output). Everything below is fresh, per-instance exploration of the stellohedron family specifically.

## n = 1 — the segment

```sage
n = 1
d = n
y = [var(f"y{i}") for i in range(1, d + 1)]
pts = reduce_codim1(stellohedron_vertices(n))
P = Polyhedron(vertices=pts)
print(f"{len(pts)} vertices:")
pts
```

### Canonical form (Proposition 6.7, general nbc method)

```sage
phi = general_canonical_form_density(P, y)
verify_pole_structure(f"Stellohedron n={n}", phi, P, y)
phi
```

### Canonical form, broken down by vertex

The stellohedron is simple, so every vertex has valency exactly d and contributes exactly one nbc term — the whole-polytope expression above already *is* essentially this sum. `canonical_form_by_vertex` (`common.sage`) is still worth running: it's the same code path used on the non-simple families elsewhere in this series (`cross_polytope_explorer.ipynb`, `hypersimplex_explorer.ipynb`, `cyclic_polytope_explorer.ipynb`), so seeing it collapse to "1 term per vertex" here is a direct, visible confirmation that the general method reduces to Proposition 6.10's one-term-per-vertex formula on a simple polytope.

```sage
rows = canonical_form_by_vertex(P, y)
print_canonical_form_by_vertex(rows)
```

### Projective dual

```sage
Dual = polar_dual(P)
print(f"dual: {Dual.n_vertices()} vertices, {Dual.n_facets()} facets")
Dual.vertices_list()
```

### Volume conjecture: canonical form vs. the volume of the projective dual (taken at the centroid)

```sage
centroid = [sum(QQ(v[i]) for v in pts) / len(pts) for i in range(d)]
pts_centered = [tuple(QQ(v[i]) - centroid[i] for i in range(d)) for v in pts]
P_centered = Polyhedron(vertices=pts_centered)

phi_centroid = general_canonical_form_density(P_centered, y)
val_at_centroid = phi_centroid.subs({yi: 0 for yi in y})

vol_dual = Dual.volume()
target = factorial(d) * vol_dual
print("phi at the centroid =", val_at_centroid)
print("d! * Vol(dual) =", target)
match_plus = bool((val_at_centroid - target) == 0)
match_minus = bool((val_at_centroid + target) == 0)
print("matches d! * Vol(dual):", match_plus, " matches -d! * Vol(dual):", match_minus)
assert match_plus or match_minus, "volume-conjecture identity failed -- would be a real bug"
```

### All triangulations, and which are regular

```sage
sp, sp_reduced, rows = secondary_polytope_data(pts)
for t, gkz, is_reg in rows:
    print(t, "GKZ vector:", gkz, " regular:", is_reg)
print(f"{len(rows)} triangulation(s) total, {sum(1 for _, _, r in rows if r)} regular")
```

### Secondary polytope: vertex embedding

```sage
print(f"secondary polytope: dimension {sp_reduced.dimension()}, {sp_reduced.n_vertices()} vertex/vertices")
sp_reduced.vertices_list()
```
